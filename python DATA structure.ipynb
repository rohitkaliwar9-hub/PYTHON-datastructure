```python
#add new data without function
myset = {10,20,30,50}
myset = myset | {40};

print(myset);
```

```python
myset = {10,20,30}
myset.add(40)
print(myset);
```

```python
#shows the unique value and same value one time only
myset1 = {10,20,30}
myset2 = {40,10,20}
print(myset1.union(myset2));
```

```python
#shows the same value onlyy
myset1 = {10,20,30}
myset2 = {40,10,20}
print(myset1.intersection(myset2));
```

```python
#shows diffrent values only elemenet the same value
myset1 = {10,20,30}
myset2 = {40,50,20}
print(myset1.symmetric_difference(myset2))
```

```python
#fetch data/value/keysfrom table
mydata = {
    'Name':'Rohit',
    'age':21,
    'salary':40000,
    'department':"CSE",
        'gender':"male",};
print(mydata.keys());
print(mydata.values());
```

```python
#update the data
mydata = {
    'Name':'Rohit',
    'age':21,
    'salary':40000,
    'department':"CSE",
        'gender':"male",};
mydata.update({'age':22});
print(mydata);
```

```python
#pop/popitem
mydata = {
    'Name':'Rohit',
    'age':21,
    'salary':40000,
    'department':"CSE",
        'gender':"male",};
print(mydata.pop('age'));
print(mydata);
mydata.popitem();
print(mydata);
```

```python
#copy the original data so if we need to modify the data the original data will be remain same
mydata = {
    'name': 'Rohit',
    'age':21,
    'salary':40000,
    'department':"CSE"};
newcopy = mydata.copy();
print(newcopy);
print(newcopy.popitem());
    
    
          
```
